## Frequências absoluta e relativa

Se $t_i$ é o token na posição $i$, a frequência de uma palavra $w$ é:

$$
f(w)=\sum_{i=1}^{N}\mathbf{1}(t_i=w),
\qquad
p(w)=\frac{f(w)}{N}.
$$

O valor de $p(w)$ só é interpretável quando $N$ está declarado. Neste
experimento mostraremos dois denominadores:

| Medida | Denominador |
|---|---|
| frequência relativa entre todos os tokens | $N$, antes de retirar stopwords |
| frequência relativa entre tokens de conteúdo | $N_c$, depois de retirar stopwords |

Remover stopwords apenas da tabela, mas manter $N$ como denominador, responde a
uma pergunta diferente de recalcular a proporção dentro do vocabulário filtrado.

```python
stopwords = {"a", "o", "e", "de", "do", "da", "como", "em", "nas", "um", "uma", "também"}
frequencias = Counter(todos_tokens)
tokens_conteudo = [token for token in todos_tokens if token not in stopwords]
frequencias_conteudo = Counter(tokens_conteudo)
total_tokens = len(todos_tokens)
total_tokens_conteudo = len(tokens_conteudo)

tabela_frequencias = pd.DataFrame([
    {
        "palavra": palavra,
        "frequencia": frequencia,
        "relativa_todos_tokens": frequencia / total_tokens,
        "relativa_tokens_conteudo": frequencia / total_tokens_conteudo,
    }
    for palavra, frequencia in frequencias_conteudo.most_common(12)
])
tabela_frequencias
```

As frequências mostram quanto um termo aparece, mas retiram as ocorrências de
seus contextos. Para avaliar o sentido e a relevância de uma contagem, precisamos
retornar aos documentos que a produziu.

## Concordâncias

Uma concordância recupera uma janela de $j$ tokens à esquerda e à direita de
cada ocorrência. Não é necessário transformar essa operação em uma medida única:
seu papel é devolver contexto ao agregado. As janelas devem respeitar as fronteiras
dos documentos e conservar o identificador da fonte.

```python
def concordancias(tabela, alvo, janela=4):
    resultados = []
    for _, documento in tabela.iterrows():
        tokens = documento["tokens"]
        for posicao, token in enumerate(tokens):
            if token == alvo:
                inicio = max(0, posicao - janela)
                fim = min(len(tokens), posicao + janela + 1)
                resultados.append({
                    "id_documento": documento["id_documento"],
                    "tema": documento["tema"],
                    "genero": documento["genero"],
                    "contexto": " ".join(tokens[inicio:fim]),
                })
    return pd.DataFrame(resultados)

concordancias(dados, "trabalho", janela=4).head(8)
```

As concordâncias permitem examinar manualmente o entorno de uma palavra. O
passo seguinte sistematiza uma pergunta relacionada: **quais palavras aparecem
juntas mais vezes do que esperaríamos apenas porque cada uma delas é frequente?**

## N-gramas e colocações

![A frequência do bigrama é comparada às frequências marginais; o resultado deve ser examinado com frequência mínima e concordâncias.](imagens/02_anatomia_pmi.svg)

Um **bigrama** é um par de tokens adjacentes, $(t_i,t_{i+1})$, dentro do mesmo
documento. Sua frequência informa quantas vezes o par foi observado, mas isso não
basta para dizer se há uma associação especialmente característica entre as duas
palavras.

Considere dois casos. Se `de` e `trabalho` aparecem muitas vezes no corpus,
`de trabalho` pode ser frequente simplesmente porque `de` é muito frequente. Já
um par como `jornada de` pode chamar atenção mesmo com menos ocorrências, caso
seus componentes apareçam juntos em proporção muito maior do que esperaríamos a
partir de suas frequências individuais.

A **informação mútua pontual** (*Pointwise Mutual Information*, PMI) formaliza
essa comparação:

$$
PMI(a,b)=\log_2\left(\frac{P(a,b)}{P_L(a)P_R(b)}\right)
=\log_2\left(\frac{c(a,b)\,N_b}{c_L(a)c_R(b)}\right).
$$

| Símbolo | Significado |
|---|---|
| $c(a,b)$ | número de ocorrências do bigrama $(a,b)$ |
| $N_b$ | número total de bigramas considerados |
| $c_L(a)$ | vezes em que $a$ aparece como primeiro elemento de um bigrama |
| $c_R(b)$ | vezes em que $b$ aparece como segundo elemento de um bigrama |

A ideia central está no quociente dentro do logaritmo. O numerador representa a
**coocorrência observada**; o denominador representa a coocorrência que
esperaríamos se as posições de $a$ e $b$ fossem independentes, dadas suas
frequências marginais.

Isso fornece uma interpretação particularmente simples:

- **PMI = 0**: o par ocorre aproximadamente na frequência esperada pelas marginais;
- **PMI > 0**: as palavras aparecem juntas **mais** do que o esperado;
- **PMI < 0**: aparecem juntas **menos** do que o esperado;
- como usamos $\log_2$, **PMI = 1** corresponde a uma coocorrência duas vezes a
  esperada; **PMI = 2**, quatro vezes; **PMI = 3**, oito vezes.

Portanto, **frequência e PMI respondem a perguntas diferentes**. Frequência
pergunta *“quais pares aparecem mais vezes?”*; PMI pergunta *“quais pares são
especialmente associados, levando em conta quão frequentes são seus componentes?”*

Há uma cautela importante. Um bigrama observado uma única vez pode receber PMI
muito alto se suas duas palavras também forem raras. O valor seria matematicamente
correto, mas pouco robusto para interpretação. Por isso, abaixo calcularemos PMI
somente para bigramas com pelo menos três ocorrências.

Finalmente, PMI é uma medida de **associação distribucional**, não de importância
histórica, significado ou causalidade. Depois de localizar um bigrama interessante,
devemos voltar aos documentos e às concordâncias para verificar em que contextos
ele efetivamente ocorre.

```python
bigramas = Counter()
marginal_esquerda = Counter()
marginal_direita = Counter()

for tokens_documento in dados["tokens"]:
    pares_documento = list(zip(tokens_documento, tokens_documento[1:]))
    bigramas.update(pares_documento)
    marginal_esquerda.update(a for a, _ in pares_documento)
    marginal_direita.update(b for _, b in pares_documento)

total_bigramas = sum(bigramas.values())
frequencia_minima = 3
linhas_pmi = []
for (a, b), frequencia in bigramas.items():
    if frequencia >= frequencia_minima:
        pmi = math.log2(
            frequencia * total_bigramas
            / (marginal_esquerda[a] * marginal_direita[b])
        )
        linhas_pmi.append((f"{a} {b}", frequencia, pmi))

tabela_colocacoes = pd.DataFrame(
    sorted(linhas_pmi, key=lambda linha: linha[2], reverse=True),
    columns=["bigrama", "frequencia", "pmi"],
)
tabela_colocacoes.head(10)
```

N-gramas e PMI observam relações locais entre palavras. Agora mudaremos a
escala: em vez de pares de tokens, compararemos o repertório lexical de documentos
inteiros.

## Vocabulário e diversidade lexical

Uma primeira distinção é entre **tokens** e **formas** (*types*).

Considere a sequência:

`trabalho salário trabalho jornada`

Ela contém quatro **tokens**, pois há quatro ocorrências de palavras, mas apenas
três **formas distintas**: `trabalho`, `salário` e `jornada`.

Se $N_d$ é o número total de tokens do documento $d$ e $V_d$ o número de formas
distintas, a **razão forma–token** (*Type-Token Ratio*, TTR) é

$$
TTR(d)=\frac{V_d}{N_d}.
$$

A TTR varia entre 0 e 1. Quanto mais próximo de 1, maior é a proporção de formas
diferentes no trecho analisado. Por exemplo:

- `trabalho salário oficina jornada` tem $TTR=4/4=1$;
- `trabalho trabalho trabalho jornada` tem $TTR=2/4=0{,}5$.

É tentador interpretar a TTR simplesmente como “riqueza de vocabulário”, mas há
um problema fundamental: **ela depende fortemente do tamanho do texto**.

Quando começamos a ler um documento, muitas palavras ainda não apareceram e é
fácil encontrar novas formas. À medida que o texto cresce, palavras já observadas
tendem a reaparecer. O denominador $N_d$ continua aumentando a cada token, enquanto
o número de formas distintas $V_d$ cresce mais lentamente. Assim, mesmo dois
textos produzidos por processos lexicais semelhantes podem apresentar TTRs
diferentes apenas porque um deles é maior.

Por isso, comparar diretamente a TTR de uma carta curta com a de um editorial
muito longo pode levar a uma conclusão enganosa: uma TTR menor no editorial **não
implica, por si só, vocabulário menos diverso**.

### Controlando o tamanho

Uma solução didática simples é comparar trechos com o mesmo número de tokens.
Escolhemos um tamanho $m$ e calculamos

$$
TTR_m(d)=
\frac{\left|\{t_1,\ldots,t_m\}\right|}{m},
\qquad N_d\geq m.
$$

Aqui adotaremos como $m$ o tamanho do menor documento. Dessa forma, todos os 24
documentos entram na comparação e todos contribuem com exatamente o mesmo número
de tokens. O gráfico seguinte permite observar lado a lado:

1. **TTR bruta**, calculada sobre o documento inteiro;
2. **TTR padronizada**, calculada sobre os primeiros $m$ tokens.

Se a relação entre tamanho e TTR enfraquecer depois da padronização, teremos uma
evidência concreta de que parte da diferença observada na TTR bruta era produzida
pelo próprio tamanho dos documentos.

A padronização, contudo, não torna a TTR uma medida perfeita. Usar os primeiros
$m$ tokens torna o resultado sensível à posição escolhida: introduções podem ter
vocabulário diferente do restante do texto. Em uma pesquisa real, poderíamos
comparar várias janelas ou amostras de tamanho fixo e registrar explicitamente o
protocolo utilizado.

Portanto, nesta aula, a TTR deve ser entendida como uma **medida simples da
proporção entre formas e tokens sob um determinado procedimento de amostragem**,
e não como uma medida absoluta de “riqueza”, qualidade ou complexidade lexical.

```python
tamanho_padrao = int(dados["tokens"].map(len).min())

def ttr_padronizada(tokens, tamanho):
    if len(tokens) < tamanho:
        return pd.NA
    segmento = tokens[:tamanho]
    return len(set(segmento)) / tamanho

diversidade = pd.DataFrame({
    "id_documento": dados["id_documento"],
    "tokens": dados["tokens"].map(len),
    "formas": dados["tokens"].map(lambda tokens: len(set(tokens))),
    "ttr": dados["tokens"].map(lambda tokens: len(set(tokens)) / len(tokens)),
    f"ttr_{tamanho_padrao}": dados["tokens"].map(
        lambda tokens: ttr_padronizada(tokens, tamanho_padrao)
    ),
})
print("Tamanho comum adotado:", tamanho_padrao, "tokens")

display(ttr_duplo(
    diversidade["tokens"].tolist(),
    diversidade["ttr"].tolist(),
    diversidade[f"ttr_{tamanho_padrao}"].tolist(),
    tamanho_padrao,
))
diversidade.head()
```

## Da contagem à representação vetorial: TF-IDF

Até aqui, usamos frequências para responder perguntas como *“quais palavras aparecem mais?”*.
Mas uma palavra muito frequente em **todos** os documentos pode ser pouco útil para distinguir
um documento de outro.

O **TF-IDF** combina duas ideias:

1. **TF — term frequency**: quanto um termo aparece em um documento;
2. **IDF — inverse document frequency**: quão raro esse termo é no conjunto de documentos.

Uma forma comum é

$$
TFIDF(t,d)=TF(t,d)\times IDF(t),
$$

com

$$
IDF(t)=\log\left(\frac{D}{DF(t)}\right),
$$

onde $D$ é o número de documentos e $DF(t)$ é o número de documentos que contêm o termo.

A consequência é importante: um termo recebe peso alto quando é frequente em um documento,
mas não aparece indiscriminadamente em todos os demais.

Isso muda a pergunta. A frequência simples descreve o corpus; TF-IDF ajuda a caracterizar
**como um documento se distingue dos outros**.

Além disso, surge algo novo: cada documento pode ser representado por uma sequência de números,

$$
\mathbf d=(w_1,w_2,\ldots,w_p),
$$

um peso para cada termo do vocabulário. Temos, portanto, uma primeira **representação vetorial
de texto**, por meio de uma operação algébrica.

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

vetorizador_tfidf = TfidfVectorizer(
    lowercase=True,
    token_pattern=r"(?u)\b[a-záàâãéêíóôõúç]+\b",
    stop_words=list(stopwords),
)

X_tfidf = vetorizador_tfidf.fit_transform(dados["texto"])
termos_tfidf = np.array(vetorizador_tfidf.get_feature_names_out())

def termos_representativos(indice_documento, n=8):
    pesos = X_tfidf[indice_documento].toarray().ravel()
    ordem = pesos.argsort()[::-1]
    ordem = [i for i in ordem if pesos[i] > 0][:n]
    return pd.DataFrame({
        "termo": termos_tfidf[ordem],
        "tf_idf": pesos[ordem],
    })

indice = 0
print(
    dados.loc[indice, ["id_documento", "tema", "genero"]].to_dict()
)
termos_representativos(indice)
```

Observe que o vetor TF-IDF não contém uma posição chamada “significado”. Cada dimensão corresponde
a um termo. Mesmo assim, operações algébricas sobre esses vetores começam a produzir resultados
interpretáveis.

Por exemplo, documentos com distribuições de termos semelhantes apontam aproximadamente para
direções semelhantes no espaço vetorial. Podemos medir o ângulo entre eles pelo **cosseno**:

$$
\cos(\mathbf x,\mathbf y)
=
\frac{\mathbf x\cdot\mathbf y}
{\|\mathbf x\|\,\|\mathbf y\|}.
$$

Valores próximos de 1 indicam vetores com direções semelhantes. Isso nos permite procurar,
por exemplo, os documentos lexicalmente mais semelhantes a um documento escolhido.

```python
from sklearn.metrics.pairwise import cosine_similarity

similaridades = cosine_similarity(X_tfidf)

indice = 0
ordem = np.argsort(similaridades[indice])[::-1]
ordem = [i for i in ordem if i != indice][:5]

pd.DataFrame({
    "id_documento": dados.loc[ordem, "id_documento"].values,
    "tema": dados.loc[ordem, "tema"].values,
    "genero": dados.loc[ordem, "genero"].values,
    "similaridade": similaridades[indice, ordem],
})
```

## De vetores esparsos a embutimentos

O TF-IDF produz vetores úteis, mas de dimensão alta: há uma coordenada para cada termo do
vocabulário. Além disso, duas palavras diferentes ocupam dimensões diferentes, mesmo quando
aparecem em contextos semelhantes.

Podemos dar outro passo usando a mesma ideia de coocorrência estudada com PMI.

Em vez de perguntar apenas se um par $(a,b)$ tem PMI alto, construímos uma **matriz
palavra–contexto**. Cada linha representa uma palavra; cada coluna, uma palavra que pode aparecer
em sua vizinhança. A célula registra a força da associação.

Usaremos **PPMI** (*Positive PMI*):

$$
PPMI(a,b)=\max(PMI(a,b),0).
$$

Assim, relações positivas são mantidas e valores negativos são substituídos por zero.

Cada linha da matriz já é um vetor que caracteriza uma palavra por seus contextos. Palavras
utilizadas em contextos parecidos tendem a ter vetores parecidos. Essa é uma forma simples de
**representação distribucional de significado**, construída apenas com contagens e álgebra.

```python
# Vocabulário de conteúdo com frequência mínima para manter a demonstração legível.
frequencia_min_palavra = 3
vocabulario = sorted([
    palavra for palavra, f in frequencias_conteudo.items()
    if f >= frequencia_min_palavra
])
indice_palavra = {p: i for i, p in enumerate(vocabulario)}

# Coocorrência em janela simétrica de duas posições.
C = np.zeros((len(vocabulario), len(vocabulario)), dtype=float)
janela = 2

for tokens in dados["tokens"]:
    tokens = [t for t in tokens if t in indice_palavra]
    for i, alvo in enumerate(tokens):
        inicio = max(0, i - janela)
        fim = min(len(tokens), i + janela + 1)
        for j in range(inicio, fim):
            if i != j:
                C[indice_palavra[alvo], indice_palavra[tokens[j]]] += 1

total = C.sum()
soma_linhas = C.sum(axis=1, keepdims=True)
soma_colunas = C.sum(axis=0, keepdims=True)

esperado = (soma_linhas @ soma_colunas) / total

with np.errstate(divide="ignore", invalid="ignore"):
    PMI = np.log2((C * total) / (soma_linhas @ soma_colunas))
    PPMI = np.where(np.isfinite(PMI) & (PMI > 0), PMI, 0)

print("Dimensão da matriz PPMI:", PPMI.shape)
```

A matriz PPMI ainda pode ter muitas dimensões. Podemos condensá-la usando uma técnica clássica
de álgebra linear: a **decomposição em valores singulares** (SVD),

$$
M = U\Sigma V^\top.
$$

Mantendo apenas os $k$ componentes mais importantes,

$$
M \approx U_k\Sigma_kV_k^\top,
$$

obtemos vetores menores:

$$
E=U_k\Sigma_k.
$$

Cada palavra passa a ser representada por poucas coordenadas. Esse vetor compacto é um
**embutimento (embedding)**.

Neste exemplo, os embutimentos surgem diretamente de coocorrências, PPMI e redução
algébrica de dimensionalidade.

```python
from sklearn.decomposition import TruncatedSVD

k = min(20, PPMI.shape[1] - 1)
svd = TruncatedSVD(n_components=k, random_state=42)
embeddings_matriz = svd.fit_transform(PPMI)

embedding = {
    palavra: embeddings_matriz[i]
    for palavra, i in indice_palavra.items()
}

def cosseno(a, b):
    denominador = np.linalg.norm(a) * np.linalg.norm(b)
    return np.dot(a, b) / denominador if denominador else 0.0

def vizinhas(palavra, n=8):
    if palavra not in embedding:
        raise ValueError(f"{palavra!r} não está no vocabulário.")
    resultados = []
    for outra, vetor in embedding.items():
        if outra != palavra:
            resultados.append(
                (outra, cosseno(embedding[palavra], vetor))
            )
    return pd.DataFrame(
        sorted(resultados, key=lambda x: x[1], reverse=True)[:n],
        columns=["palavra", "similaridade"],
    )

vizinhas("trabalho")
```

### Álgebra sobre representações

Agora podemos realizar operações que não faziam sentido sobre palavras como símbolos isolados.

**Similaridade**

$$
\cos(\mathbf e_{\text{trabalho}},\mathbf e_{\text{jornada}})
$$

mede proximidade distribucional.

**Composição aproximada**

$$
\mathbf q=
\mathbf e_{\text{saúde}}+\mathbf e_{\text{pública}}
$$

produz uma nova direção no espaço. Podemos procurar palavras próximas desse vetor. A soma não
“compreende” a expressão; ela combina geometricamente dois padrões de contexto. Ainda assim,
o resultado pode ser semanticamente interpretável.

**Diferenças**

$$
\mathbf e_a-\mathbf e_b
$$

representam deslocamentos no espaço. Em corpora suficientemente ricos, diferenças semelhantes
podem corresponder a relações linguísticas semelhantes. Isso é uma propriedade empírica do espaço,
não uma garantia de raciocínio ou compreensão.

O ponto central é que **relações semânticas passam a ser investigadas por relações algébricas**.

```python
def mais_proximas_do_vetor(vetor, n=8, excluir=()):
    resultados = []
    for palavra, e in embedding.items():
        if palavra not in excluir:
            resultados.append((palavra, cosseno(vetor, e)))
    return pd.DataFrame(
        sorted(resultados, key=lambda x: x[1], reverse=True)[:n],
        columns=["palavra", "similaridade"],
    )

if "saúde" in embedding and "pública" in embedding:
    vetor_composto = embedding["saúde"] + embedding["pública"]
    display(
        mais_proximas_do_vetor(
            vetor_composto,
            excluir={"saúde", "pública"}
        )
    )
```

## Um problema: o significado depende do contexto

Os embutimentos construídos acima são **estáticos**: uma palavra recebe o mesmo vetor sempre que
aparece.

Mas considere uma palavra ambígua em dois contextos:

- `O banco concedeu crédito aos trabalhadores.`
- `Os moradores descansaram no banco da praça.`

Se `banco` tiver um único vetor, a representação não distingue diretamente essas duas ocorrências.
A informação que ajuda a interpretar cada caso está nas **outras palavras presentes no contexto**.

Isso sugere uma operação simples: ao construir uma representação para uma palavra, consulta ou
trecho, podemos atribuir pesos diferentes às informações disponíveis conforme sua relação com
aquilo que estamos procurando.

Chamaremos esse procedimento de **atenção**. O objetivo aqui não é apresentar um método ótimo,
mas tornar visível a ideia matemática: **comparar, ponderar e combinar informações de acordo com
o contexto**.


## Atenção: selecionar informação relevante para uma representação

Comecemos por uma versão simples, de forma inteiramente explícita.

Suponha que nossa consulta seja `saúde`. Seu embedding será o vetor de consulta,

$$
\mathbf q=\mathbf e_{\text{saúde}}.
$$

Para cada palavra $i$ de um trecho, temos um vetor $\mathbf k_i$. Calculamos a compatibilidade
pela similaridade vetorial. Em uma versão elementar:

$$
s_i=\cos(\mathbf q,\mathbf k_i).
$$

Em seguida, transformamos os escores em pesos positivos cuja soma é 1:

$$
\alpha_i=
\frac{\exp(s_i/\tau)}
{\sum_j \exp(s_j/\tau)}.
$$

Essa transformação é o **softmax**. O parâmetro $\tau$ controla quão concentrados ficam os pesos.

Finalmente, construímos uma representação contextual como soma ponderada:

$$
\mathbf z=\sum_i \alpha_i\mathbf v_i.
$$

Nesta demonstração, usaremos o próprio embedding da palavra tanto como *key* quanto como *value*.
Assim podemos observar o mecanismo sem treinar modelo algum.

```python
def softmax(x):
    x = np.asarray(x, dtype=float)
    x = x - np.max(x)
    exp_x = np.exp(x)
    return exp_x / exp_x.sum()

def atencao_simples(consulta, tokens, temperatura=0.20):
    tokens_validos = [
        t for t in tokens
        if t in embedding
    ]

    q = embedding[consulta]
    V = np.vstack([
        embedding[t]
        for t in tokens_validos
    ])

    # Compatibilidade consulta-palavra.
    escores = np.array([
        cosseno(q, v)
        for v in V
    ])

    pesos = softmax(escores / temperatura)

    # Representação do contexto orientada pela consulta.
    contexto = np.sum(
        pesos[:, None] * V,
        axis=0
    )

    tabela = pd.DataFrame({
        "token": tokens_validos,
        "compatibilidade": escores,
        "atencao": pesos,
    }).sort_values("atencao", ascending=False)

    return tabela, contexto

# Escolhemos um documento que contenha vocabulário de saúde.
candidatos = dados.index[
    dados["tokens"].map(lambda ts: "saúde" in ts)
].tolist()

indice = candidatos[0]
print(dados.loc[indice, ["id_documento", "tema", "genero"]].to_dict())

tabela_atencao, contexto_saude = atencao_simples(
    "saúde",
    dados.loc[indice, "tokens"]
)

tabela_atencao.head(15)
```

A coluna `atencao` não representa uma importância universal da palavra. Ela responde:

> **dada esta consulta e este conjunto de representações, quanto cada ocorrência contribui para
> a representação que estamos construindo?**

Mude a consulta de `saúde` para `trabalho` no mesmo documento. Os pesos podem mudar porque a
informação procurada mudou.

Essa é uma característica essencial: atenção é **relacional e contextual**, não um ranking fixo
de palavras.

```python
# Comparação de duas consultas sobre o mesmo documento.
consultas = [
    q for q in ["saúde", "trabalho"]
    if q in embedding
]

comparacao = []

for consulta in consultas:
    tabela, _ = atencao_simples(
        consulta,
        dados.loc[indice, "tokens"]
    )
    # Soma os pesos quando a mesma palavra ocorre várias vezes.
    agregado = (
        tabela.groupby("token", as_index=False)["atencao"]
        .sum()
    )
    agregado["consulta"] = consulta
    comparacao.append(agregado)

pd.concat(comparacao).sort_values(
    ["consulta", "atencao"],
    ascending=[True, False]
).groupby("consulta").head(10)
```

## Atenção como operação de representação contextual

A demonstração anterior pode ser resumida em três passos:

1. representar a consulta e as palavras por vetores;
2. calcular quanto cada palavra se relaciona com a consulta;
3. combinar os vetores das palavras usando essas relações como pesos.

Se $\mathbf q$ representa aquilo que estamos procurando e $\mathbf v_i$ representa a palavra
$i$, calculamos primeiro um escore de compatibilidade:

$$
s_i = sim(\mathbf q,\mathbf v_i).
$$

Aqui usamos similaridade por cosseno. Em seguida, o softmax transforma os escores em pesos:

$$
\alpha_i =
\frac{\exp(s_i/\tau)}
{\sum_j \exp(s_j/\tau)}.
$$

Finalmente:

$$
\mathbf z =
\sum_i \alpha_i\mathbf v_i.
$$

O vetor $\mathbf z$ é uma **combinação ponderada do contexto orientada pela consulta**.

Esse procedimento é propositalmente simples. Ele tem limitações importantes: depende da qualidade
dos embeddings, usa uma regra fixa de similaridade e não resolve sozinho ambiguidade, sintaxe ou
relações de longa distância. Sua vantagem didática é permitir observar diretamente cada operação.

### Da consulta externa ao contexto de uma palavra

Também podemos usar uma palavra do próprio trecho como consulta. Para representar `banco` em

`O banco concedeu crédito aos trabalhadores`

podemos comparar o vetor de `banco` aos vetores das demais palavras e construir uma combinação
ponderada daquele contexto. Repetindo o procedimento em

`Os moradores descansaram no banco da praça`

os pesos e a combinação resultante podem mudar.

Assim, mesmo partindo de um embedding estático, podemos construir uma representação que incorpora
informação do contexto:

$$
\mathbf z_{\text{banco, contexto 1}}
\neq
\mathbf z_{\text{banco, contexto 2}}.
$$

Não devemos concluir daí que o procedimento “descobriu o significado” de `banco`. O que ele fez
foi mais preciso e observável: **produziu representações diferentes ao combinar informações
contextuais diferentes**.

Podemos resumir o percurso da aula:

$$
\text{contagem}
\rightarrow
\text{TF-IDF}
\rightarrow
\text{coocorrência/PPMI}
\rightarrow
\text{embedding}
\rightarrow
\text{ponderação por atenção}
\rightarrow
\text{representação dependente do contexto}.
$$

Há uma mudança progressiva de pergunta:

- **frequência:** o que aparece?
- **TF-IDF:** o que caracteriza este documento em relação aos demais?
- **PPMI/embedding:** quais unidades aparecem em contextos semelhantes?
- **atenção:** quais informações disponíveis são mais relacionadas àquilo que estamos
  representando agora?

A atenção é apresentada aqui, portanto, como uma operação algébrica simples para **selecionar e
combinar informação contextual**, e não como uma tecnologia específica.

relações locais, documentos vetoriais, embeddings e atenção. A atividade final
deve explicitar o que se ganha — e o que se perde — em cada transformação.
